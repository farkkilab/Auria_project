```python
import os
import pandas as pd
from chardet import detect
```

```python
test = pd.read_csv('/Volumes/group-1/h345/afarkkilab/Data/17_Auria_TMA/csv/raw/01A_raw.csv')
```

```python

# 文件夹路径
directory = "/Volumes/group-1/h345/afarkkilab/Data/17_Auria_TMA/csv/raw"
# 目标样品名称
samples = {'19_02_02A', '19_02_01A', 'T2', 'INF2', '19_02_03A', '1B', '2B', '3B', 'INF1'}
# 目标文件格式
file_suffix = "_raw.csv"

def detect_encoding(file_path):
    """检测文件编码"""
    with open(file_path, 'rb') as f:
        raw_data = f.read(50000)  # 读取前50000字节来猜测编码
    return detect(raw_data)['encoding']

# 找到符合条件的文件
files_to_merge = []
for file in os.listdir(directory):
    if file.endswith(file_suffix):
        sample_name = file.replace(file_suffix, "")
        if sample_name in samples:
            files_to_merge.append(os.path.join(directory, file))

# 读取并合并所有 CSV（自动处理编码）
if files_to_merge:
    df_list = []
    for f in files_to_merge:
        try:
            # 先尝试 utf-8
            df = pd.read_csv(f, encoding='utf-8')
        except UnicodeDecodeError:
            # 如果失败，检测编码并重试
            file_encoding = detect_encoding(f)
            df = pd.read_csv(f, encoding=file_encoding)
        df_list.append(df)
    
    merged_df = pd.concat(df_list, ignore_index=True)
    
    # 保存合并后的文件（使用 UTF-8 编码）
    output_file = os.path.join("/Users/wenqchen/Desktop/Projects/Auria/Data/t-CycIF/raw_merged_data.csv")
    merged_df.to_csv(output_file, index=False, encoding='utf-8')
    print(f"合并完成，文件保存至: {output_file}")
else:
    print("未找到符合条件的文件。")
```
